```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        # Últimos i elementos ya están ordenados
        for j in range(0, n-i-1):
            # Intercambia si el elemento encontrado es mayor que el siguiente
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr
```

```python
def insertion_sort(arr):
    # Recorre desde el segundo elemento hasta el final
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        # Mueve los elementos de arr[0..i-1], que son mayores que key, a una posición adelante
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr
```

Tienes una lista de nombres de estudiantes en un aula, pero no están ordenados alfabéticamente. Escribe una función que utilice el algoritmo de ordenamiento por Inserción para ordenar estos nombres en orden alfabético. Una vez ordenados, imprime la lista ordenada y cuenta cuántos nombres empiezan con la letra "A".

```python
def contar_nombres_con_a(nombres):
    count = 0
    for nombre in nombres:
        if nombre.startswith("A"):
            count += 1
    return count

def ordenar_nombres_y_contar_a(nombres):
    # Ordenar los nombres usando el algoritmo de Inserción
    insertion_sort(nombres)

    # Contar cuántos nombres empiezan con 'A'
    count_a = contar_nombres_con_a(nombres)

    # Imprimir resultados
    print("Nombres ordenados:", nombres)
    print("Cantidad de nombres que empiezan con 'A':", count_a)

# Ejemplo de uso
nombres = ["Luis", "Ana", "Mario", "Andrea", "Carla", "Alberto", "Beatriz"]
ordenar_nombres_y_contar_a(nombres)
```

Imagina que tienes la siguiente lista de personas:

```python
personas = [
    {"nombre": "Luis", "edad": 34},
    {"nombre": "Ana", "edad": 29},
    {"nombre": "Mario", "edad": 40},
    {"nombre": "Ana", "edad": 22},
    {"nombre": "Luis", "edad": 25},
    {"nombre": "Beatriz", "edad": 30}
]
```

Queremos ordenar esta lista primero por el nombre alfabéticamente y, en caso de que dos nombres sean iguales, por edad en orden ascendente.

```python
def bubble_sort_personas(personas):
    n = len(personas)
    for i in range(n):
        for j in range(0, n-i-1):
            if (personas[j]["nombre"] > personas[j+1]["nombre"]) or \
               (personas[j]["nombre"] == personas[j+1]["nombre"] and personas[j]["edad"] > personas[j+1]["edad"]):
                personas[j], personas[j+1] = personas[j+1], personas[j]
    return personas

print(bubble_sort_personas(personas))
```

```python
def insertion_sort_personas(personas):
    for i in range(1, len(personas)):
        key = personas[i]
        j = i - 1
        while j >= 0 and (
            personas[j]["nombre"] > key["nombre"] or
            (personas[j]["nombre"] == key["nombre"] and personas[j]["edad"] > key["edad"])
        ):
            personas[j + 1] = personas[j]
            j -= 1
        personas[j + 1] = key
    return personas
```

```python
import time
# Datos de entrada
personas = [
    {"nombre": "Luis", "edad": 34},
    {"nombre": "Ana", "edad": 29},
    {"nombre": "Mario", "edad": 40},
    {"nombre": "Ana", "edad": 22},
    {"nombre": "Luis", "edad": 25},
    {"nombre": "Beatriz", "edad": 30}
]

# Ordenar usando Burbuja
start_time = time.time()
ordenado_burbuja = bubble_sort_personas(personas.copy())
end_time = time.time()
tiempo_burbuja = end_time - start_time
print("Ordenado por Burbuja:")
for persona in ordenado_burbuja:
    print(persona)
print(f"Tiempo de ejecución (Burbuja): {tiempo_burbuja:.6f} segundos")

# Ordenar usando Inserción
start_time = time.time()
ordenado_insercion = insertion_sort_personas(personas.copy())
end_time = time.time()
tiempo_insercion = end_time - start_time
print("\nOrdenado por Inserción:")
for persona in ordenado_insercion:
    print(persona)
print(f"Tiempo de ejecución (Inserción): {tiempo_insercion:.6f} segundos")
```

Imagina que trabajas en una cafetería donde los pedidos de los clientes llegan en orden aleatorio y cada uno tiene un tiempo de preparación estimado en minutos. Para optimizar el trabajo de los baristas, queremos ordenar los pedidos de menor a mayor tiempo de preparación utilizando ordenamiento burbuja en una lista ligada.

Planteamiento:
Implementar una lista ligada donde cada nodo representa un pedido.

Cada pedido tendrá:

ID del cliente

Tiempo de preparación estimado (en minutos)

Usar ordenamiento burbuja en la lista ligada para organizar los pedidos.

```python
class Pedido:
    def __init__(self, id_cliente, tiempo_preparacion):
        self.id_cliente = id_cliente
        self.tiempo_preparacion = tiempo_preparacion
        self.siguiente = None  # Apunta al siguiente pedido

class ListaLigadaPedidos:
    def __init__(self):
        self.cabeza = None

    def agregar_pedido(self, id_cliente, tiempo_preparacion):
        """ Agrega un pedido al final de la lista ligada """
        nuevo_pedido = Pedido(id_cliente, tiempo_preparacion)
        if not self.cabeza:
            self.cabeza = nuevo_pedido
            return
        actual = self.cabeza
        while actual.siguiente:
            actual = actual.siguiente
        actual.siguiente = nuevo_pedido

    def imprimir_pedidos(self):
        """ Muestra los pedidos en la lista ligada """
        actual = self.cabeza
        pedidos = []
        while actual:
            pedidos.append(f"Cliente {actual.id_cliente}: {actual.tiempo_preparacion} min")
            actual = actual.siguiente
        print(" -> ".join(pedidos))

    def ordenar_pedidos_burbuja(self):
        """ Ordena los pedidos en la lista ligada por tiempo de preparación usando Bubble Sort """
        if not self.cabeza or not self.cabeza.siguiente:
            return  # Lista vacía o con un solo elemento

        ordenado = False
        while not ordenado:
            actual = self.cabeza
            ordenado = True
            while actual.siguiente:
                if actual.tiempo_preparacion > actual.siguiente.tiempo_preparacion:
                    # Intercambiar valores en lugar de nodos
                    actual.tiempo_preparacion, actual.siguiente.tiempo_preparacion = actual.siguiente.tiempo_preparacion, actual.tiempo_preparacion
                    actual.id_cliente, actual.siguiente.id_cliente = actual.siguiente.id_cliente, actual.id_cliente
                    ordenado = False
                actual = actual.siguiente

# Crear la lista ligada de pedidos
cafeteria = ListaLigadaPedidos()
cafeteria.agregar_pedido(1, 10)
cafeteria.agregar_pedido(2, 5)
cafeteria.agregar_pedido(3, 7)
cafeteria.agregar_pedido(4, 3)

print("\nPedidos antes de ordenar:")
cafeteria.imprimir_pedidos()

# Ordenar los pedidos usando burbuja
cafeteria.ordenar_pedidos_burbuja()

print("\nPedidos después de ordenar:")
cafeteria.imprimir_pedidos()
```
